## Flowcharts and Planning Ahead

It's important to think about what you want your code to do ahead of time. If you write something without planning it, it might be wrong or confusing, and generally you're not going to want to go back and rewrite the whole thing. This can lead to code that is bloated, inefficient, or just confusing to read!

So plan it out a bit. You might do that with a flowchart, mapping out precise decision points and their routes. You might do this with 'pseudocode', where you write code-like but not in an actual language. Or you might do this with a notepad, a bunch of arrows and some words. Whatever helps you think ahead.
- 'This leads to this, leads to that, maybe back to this'
- 'if X, then Y. Otherwise, do Z. Z: for each character, count it and store that in an array. Then...'

The website draw.io will allow you to make proper visual flowcharts; even if you don't *need* it for this, give it a go!
